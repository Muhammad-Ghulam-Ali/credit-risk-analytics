```python
import sys
sys.path.append('..')
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from scipy.stats import ttest_ind
```

```python
# Load Data
df = pd.read_csv('../data/processed/clean_data.csv')
df.head()
```

```python
# Two-sample t-test: compare AMT_INCOME_TOTAL between defaulters (TARGET=1) and non-defaulters (TARGET=0)
# Print the t-statistic and p-value, interpret whether the difference is statistically significant at alpha=0.05

df['AGE'] = ((df['DAYS_BIRTH'] * -1) / 365).round().astype(int)

defaulters = df[df['TARGET'] == 1]
non_defaulters = df[df['TARGET'] == 0]

defaulters_income = defaulters['AMT_INCOME_TOTAL']
non_defaulters_income = non_defaulters['AMT_INCOME_TOTAL']

t_stat, p_value = stats.ttest_ind(defaulters_income, non_defaulters_income)

print("=" * 30)
print("T TEST")
print(" " * 30)
print(f't-statistics:        {t_stat.round(2)}')
print(f'p_value:        {p_value.round(2)}')

print(" " * 30)
print("=" * 30)
print("INTERPRETATION:")
print("""Since p-value (0.03) < alpha (0.05), we reject the null hypothesis. 
There is a statistically significant difference in AMT_INCOME_TOTAL 
between defaulters and non-defaulters, with defaulters having a lower 
average income (t-statistic = -2.21).
""")
```

```python
# Two-sample t-test: compare AGE (in years) between defaulters and non-defaulters
# Print t-statistic and p-value, interpret significance and direction

defaulters_age = defaulters['AGE']
non_defaulters_age = non_defaulters['AGE']

t_test,p_value = stats.ttest_ind(defaulters_age, non_defaulters_age)

print("=" * 30)
print("T TEST")
print(" " * 30)
print(f't-statistics:        {t_test.round(2)}')
print(f'p_value:              {p_value.round(2)}')

print(" " * 30)
print("=" * 30)
print("INTERPRETATION:")
print("""Since p-value (0.0) < alpha (0.05), we reject the null hypothesis. 
There is a statistically significant difference in AGE between 
defaulters and non-defaulters, with defaulters being younger on 
average (t-statistic = -43.53).""")
```

```python
# Chi-square test: CODE_GENDER vs TARGET
# Build a contingency table (crosstab), run chi-square test, print chi2 statistic and p-value, interpret
contingency_table = pd.crosstab(df['CODE_GENDER'], df['TARGET'])
chi2_stat, p_value, dof, expected = stats.chi2_contingency(contingency_table)

print("=" * 30)
print("CHI-SQUARE TEST")
print(" " * 30)
print(f'chi2-statistics:        {chi2_stat.round(2)}')
print(f'p_value:                 {p_value.round(2)}')

print(" " * 30)
print("=" * 30)
print("INTERPRETATION:")
print("""Since p-value (0.0) < alpha (0.05), we reject the null hypothesis. 
There is a statistically significant association between 
CODE_GENDER and TARGET, meaning default status is not 
independent of gender (chi2-statistic = 920.79).""")
```
